```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/Kylejunker/Semester-Group-Project/refs/heads/main/teams_hitting.csv"
df = pd.read_csv(url)

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

plt.figure(figsize=(10, 6))
plt.scatter(df['BA'], df['HR'], alpha=0.5)
plt.xlabel('Batting Average (BA)')
plt.ylabel('Homeruns (HR)')
plt.title('Batting Average vs Homeruns')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/Kylejunker/Semester-Group-Project/refs/heads/main/Team_pitching.csv"
df = pd.read_csv(url)

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

plt.figure(figsize=(10, 6))
plt.scatter(df['ERA'], df['W-L%'], alpha=0.5)
plt.xlabel('Earned Run Average (ERA)')
plt.ylabel('Win Lose % (W-L%)')
plt.title('Earned Run Average vs Win Lose %')
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

url = "https://raw.githubusercontent.com/Kylejunker/Semester-Group-Project/refs/heads/main/JudgeVOhtani.csv"
df = pd.read_csv(url)

plt.figure(figsize=(12, 6))
plt.bar(df['Player'], df['HR'], color='steelblue')
plt.xlabel('Player')
plt.ylabel('Homerun')
plt.title(f'Comparison of Aaron Judge and Shohei Ohtani - HR')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
plt.bar(df['Player'], df['BA'], color='steelblue')
plt.xlabel('Player')
plt.ylabel('Batting Average')
plt.title(f'Comparison of Aaron Judge and Shohei Ohtani - BA')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
plt.bar(df['Player'], df['SO'], color='steelblue')
plt.xlabel('Player')
plt.ylabel('Strike Outs')
plt.title(f'Comparison of Aaron Judge and Shohei Ohtani - SO')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


plt.figure(figsize=(12, 6))
plt.bar(df['Player'], df['BB'], color='steelblue')
plt.xlabel('Player')
plt.ylabel('Walks')
plt.title(f'Comparison of Aaron Judge and Shohei Ohtani - BB')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
plt.bar(df['Player'], df['OPS'], color='steelblue')
plt.xlabel('Player')
plt.ylabel('OPS')
plt.title(f'Comparison of Aaron Judge and Shohei Ohtani - OPS')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
plt.bar(df['Player'], df['OPS+'], color='steelblue')
plt.xlabel('Player')
plt.ylabel('OPS+')
plt.title(f'Comparison of Aaron Judge and Shohei Ohtani - OPS+')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
